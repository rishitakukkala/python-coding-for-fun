```python
def consecutive_sum(n: int) -> bool:
    if n < 3:
        return False

    m = 2
    while m * (m + 1) <= 2 * n:
        numerator = 2* n - m * (m-1)
        if numerator % (2 * m) ==0:
            k = numerator // (2 *m)
            if k > 0:
                return True
        m += 1
    return False
    
```

```python
consecutive_sum(100000000)
```
