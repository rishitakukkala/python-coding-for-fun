Given a non-negative integer numRows, generate the first numRows of Pascal's triangle.


In Pascal's triangle, each number is the sum of the two numbers directly above it.

Example:

Input: 5
Output:
[
     [1],
    [1,1],
   [1,2,1],
  [1,3,3,1],
 [1,4,6,4,1]
]

```python
class Solution:
    def helper(self,rowNums):
        triangle = [[1]*i for i in range(1,rowNums+1)]
        
        rows = 1
        while(rows<rowNums):
            for cols in range(1,len(triangle[rows-1])):
                triangle[rows][cols] = triangle[rows-1][cols]+triangle[rows-1][cols-1]
            rows+= 1
        return triangle
        
        
        
sol = Solution()
sol.helper(5)
```
